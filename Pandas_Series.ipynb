```python
# creating Series using Dictionary
fruit_calories = {
    "Apple": 52,
    "Banana": 89,
    "Orange": 47,
    "Strawberry": 32,
    "Grapes": 69,
    "Watermelon": 30,
    "Mango": 60,
    "Pineapple": 50,
    "Avocado": 160, # Higher in healthy fats
    "Kiwi": 61
}
s2 = pd.Series(fruit_calories,name = "calories")
s2
```

```python
#Conditionl Selection:
s2[s2>40]
```

```python
# Logical operators: and,or,not
# and ---> &
s2[(s2>40) & (s2<70)]
```

```python
# or ---> |
s2[(s2>40) | (s2<70)]
```

```python
# not ---> ~
s2[~(s2 > 40)]
```

```python
# modification in Series
s2['Apple'] = 100
s2
```
